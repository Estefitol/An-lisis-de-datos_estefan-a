### Gráfica base

Para crear un gráfico en ggplot2 se utiliza la función ggplot() para establecer un objeto de tipo gráfico al que se le agregarán
diferentes capas.
El primer parámetro de ggplot() corresponde al conjunto de datos que se utilizará en el gráfico; sin embargo, si solo integramos éste parámetro, el gráfico no muestra nada, ya que aún no se indica cómo visualizar los datos.
Podemos considerar este gráfico inicial como un gráfico base en la que se irán añadiendo capas para diseñar la visualización

```r
ggplot (data = grupos)
```

### Mapeo

El argumento mapping de la función gglopt() define cómo se asignan las variables del conjunto de datos a las propiedades visuales del gráfico. Utiliza la función aes () para definir el mapeo, donde los argumentos x e y especifican qué variables asignar a los ejes x e y respectivamente; se debe verificar que los nombres de las variables coincidan con los del conjunto de datos.

```r
ggplot(data = grupos, mapping = aes (x = estatura, y = calzado))
```

#### Objetos geométricos

El diseño del gráfico va avanzando, pero aún no aparecen los datos; esto se debe a que no le hemos indicado a R cómo representar las observaciones.
Para representar los datos del gráfico, debemos definir una "geom", que es el objeto geométrico utilizando para representar los datos en ggplot2. Cada tipo de gráfico utiliza una geom espeífica. Por ejemplo:

  - Geom_point() : Agrega una capa de puntos y crea un gráfico de dispersión
  - Geom_bar(): Crea una gráfica de barras
  - Geom_line(): Crea una gráfica de líneas
  - Geom_boxplot(): Permite diseñar un diagrama de caja y brazos.

```r
ggplot(data = grupos, mapping = aes(x = estatura, y = calzado)) + geom_point()
```

### Mapas estéticos
 Incorporemos una tercera variable al gráfico para representarla mediante puntos de diferentes colores. Para lograr esto, necesita modificar el mapeo estetico dentro de la función aes().

### Gráficas puntos

:::Cada observación se dibuja como un punto, y los puntos que comparten valor se apilan. La altura de la pila es la frecuencia de esa categoría. :::

**Cuándo usarla**: con pocos casos, cuando interesa que se vea que detrás de cada unidad hay una observación. Con muchos casos deja de leerse y conviene la barra.

```r
# gráfica de puntos

ggplot (data = grupos, mapping = aes (x = carrera)) +
  geom_dotplot (binaxis = "x", stackdir = "up", dotsize = 0.5, fill = azul)
```

```r

ggplot (data = grupos, mapping = aes (x = str_wrap(carrera, 18))) +
  geom_dotplot (binaxis = "x", stackdir = "up", dotsize = 0.5, fill = azul) +
  scale_y_continuous (NULL, breaks = NULL) +
  labs (title = "Un punto por cada integrante del equipo", x = NULL)
```

### Gráfica de barras

::: Representa la frecuencia de cada categoría con la longitud de una barra. Todas las barras arrancan del mismo punto, así que el ojo compara los extremos. :::

**Cuándo usarla**:para una variable categórica, siempre. Es la opción por defecto y la más fácil de leer con precisión.
Se construye por capas. Primero la básica:

```r
ggplot(data = grupos, mapping = aes (x = carrera)) +
  geom_bar()
```

```r
# Mejora de la gráfica de barras
ggplot(data = grupos, mapping = aes (y = carrera)) +
 geom_bar(fill = tinta) +
  labs(x = "Estudiantes", y = NULL)
```

### Ordenar las categorías

ggplot ordena alfabéticamente, que casi nunca es el orden que interesa. Para ordenar por frecuencia se convierte la variable en factor.
Un factor es una variable categórica con sus niveles declarados y en un orden definido. fct_infreq() ordena por frecuencia;
fct_rev() invierte.

```r
ggplot ( data = grupos, mapping = aes (y =  fct_rev (fct_infreq(carrera)))) +
  geom_bar(fill = tinta) +
  labs (x = "Estudiantes", y = NULL)
```

### Etiquetar los valores

Cuando importa la cifra exacta, se escribe sobre la barra y se puede quitar el eje, que ya no aporta.

```r
ggplot(data = grupos, mapping = aes(y = fct_rev(fct_infreq(carrera)))) +
geom_bar(fill = tinta) +
geom_text(stat = "count", aes(label = after_stat(count)),
          hjust = 1.4, #posición horizontal de la etiqueta
          color = "white",
          size = 4
          ) +
labs(x = "Estudiantes", y = NULL)
```
