1. **Quién levanta el dato y con qué propósito:** El Banco Mundial recopila estos datos a través de fuentes estadísticas oficiales de los países para medir el desarrollo económico, rastrear el crecimiento y diseñar políticas globales de reducción de la pobreza.

## Escala de medición de cada variable
* **País (`country.value`):** Variable categórica con escala **nominal**.
* **Fecha o Año (`date`):** Variable numérica con escala de **intervalo** (o razón, dependiendo del enfoque temporal).
* **PIB en USD (`value`):** Variable numérica continua con escala de **razón** (el cero absoluto significa ausencia de producción).

## Las cuatro preguntas contestadas
1. **¿Qué quiero saber?** Quiero entender la evolución y tendencia del Producto Interno Bruto de México a lo largo de los años recientes.
2. **¿Qué datos necesito?** Necesito una serie de tiempo anual que registre el valor total de la economía mexicana (PIB en dólares estadounidenses).
3. **¿De dónde los saco?** De la API pública de indicadores del Banco Mundial.
4. **¿Qué voy a hacer con ellos?** Voy a importarlos a R, limpiar la tabla para quedarme con las variables clave (país, año, valor) y analizar la distribución y medidas de tendencia central de la serie.

```r
library(tidyverse)
```

```r
library(jsonlite)
```

```r
url_banco_mundial <- "https://api.worldbank.org/v2/country/MEX/indicator/NY.GDP.MKTP.CD?format=json"
```

```r
grupos <- fromJSON(url_banco_mundial)
```

```r
grupos
```

```r
datos_mexico <- grupos [[2]]
```

```r
datos_limpios <- datos_mexico %>%
  mutate(
    pais = country$value,
    año = as.numeric(date)
  ) %>%
  select(pais, año, pib_usd = value) %>%
  drop_na(pib_usd) %>%
  arrange(año)

head(datos_limpios)
```
