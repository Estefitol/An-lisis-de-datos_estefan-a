## 3 · Leer la base

La función depende de la extensión del archivo. La tabla completa y los detalles
de cada formato están en la
[referencia de R](https://cjjmdata.github.io/analisis_datos_i/curso/referencia-r.html#abrir-un-archivo-de-datos).

Guarda la base en un objeto llamado **`serie`**. Es el nombre que usan todas las
sesiones del curso para tu base.

Tres situaciones frecuentes:

**El archivo está en una dirección de internet y es `.csv` o `.csv.gz`**

```r
serie <- read_csv("https://direccion/del/archivo.csv.gz")
```

**El archivo está en internet y es Excel, SPSS o un `.zip`**: primero se descarga
a la máquina de Colab.

```r
download.file("https://direccion/del/archivo.zip", "archivo.zip", mode = "wb")
unzip("archivo.zip")
list.files(recursive = TRUE)      # qué archivos salieron del zip, también dentro de carpetas
serie <- read_csv("nombre_del_archivo.csv")
```

**Descargaste el archivo a tu computadora**: súbelo con el icono de carpeta de la
barra izquierda, y léelo por su nombre.

```r
serie <- read_excel("archivo.xlsx")
```

Escribe tu lectura en la celda de abajo.

````r
# Lectura de la base
download.file("https://api.worldbank.org/v2/en/indicator/NY.GDP.MKTP.CD?downloadformat=excel", "pib_mexico.xls", mode = "wb")
serie <- read_excel("pib_mexico.xls", skip = 3)
````

### Variables

Elige de **3 a 5 variables**, incluidas las que nombraste en tu pregunta.

El **significado** sale del diccionario de datos, del descriptor de archivos o de
la página de la fuente, no del nombre de la columna. Escribe de dónde lo sacaste.

El **tipo en R** es el que aparece entre `< >` en la salida de `glimpse()`.
El **tipo estadístico** es el que vimos en la sesión 4: categórica nominal,
categórica ordinal, numérica discreta o numérica continua.

| Variable | Qué significa | Tipo en R | Tipo estadístico |
|---|---|---|---|
| Country name | Nombre del país o región | <chr> | Categórica nominal |
| Country Code| Código ISO del identificacion del país | <chr> | Categórica nominal |
| 2024 | Valor del Producto Interno Bruto en dólares estadounidenses | <dbl> | Númerica continua |

*Fuente del significado de las variables: Metadatos oficiales y documentación del Banco Mundial.*

### Tipo en R y tipo estadístico

R asigna el tipo según lo que encuentra en el archivo, no según lo que la variable
significa. Una clave numérica que representa categorías puede llegar como `<dbl>`.

Sí coinciden  porque las columnas de los años contienen montos económicos y R las asigna correctamente como "dbl", cosa que significa que corresponde perfectamente con una escala numérica continua.
